```python
# Exercise 1

# === REQUIRED VARIABLES ===
SEED_NUM = 5  # <-- CHANGE THIS
STUDENT_MAJOR = "BSECE"  # MUST BE UPPERCASE
FAVORITE_ARTIST = "BEN&BEN"
LOCAL_HAZARD = "FLOOD"

CONTROL_NUM = max(3, SEED_NUM)

# === EXERCISE 1 ===
from collections import defaultdict, Counter

# Base stream
stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

# Data Expansion (ADD EXACTLY CONTROL_NUM + 3)
custom_entries = []
for i in range(CONTROL_NUM + 3):
    campus = "TUP Manila" if i % 2 == 0 else "TUP Taguig"
    program = f"{STUDENT_MAJOR}_v{i}"
    custom_entries.append((campus, program))

stream.extend(custom_entries)

# Phase 1: Aggregation
campus_data = defaultdict(list)

for campus, program in stream:
    campus_data[campus].append(program)

# Phase 2: Frequency Analysis
manila_counter = Counter(campus_data["TUP Manila"])
top_program, top_count = manila_counter.most_common(1)[0]

# Outputs
print("=== EXERCISE 1 OUTPUT ===")
print("CONTROL_NUM:", CONTROL_NUM)
print("STUDENT_MAJOR:", STUDENT_MAJOR)
print("Expanded Stream Length:", len(stream))
print("TUP Manila Applications:", len(campus_data["TUP Manila"]))
print("Top Program (Manila):", top_program)
print("Frequency:", top_count)
```
