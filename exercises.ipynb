```python
# EXERCISE 1

# === REQUIRED VARIABLES ===
SEED_NUM = 1  # <-- CHANGE THIS
STUDENT_MAJOR = "BSME"  # MUST BE UPPERCASE
FAVORITE_ARTIST = "NICOLE"
LOCAL_HAZARD = "TYPHOON"

CONTROL_NUM = max(3, SEED_NUM)

# === EXERCISE 1 ===
from collections import defaultdict, Counter

# Base stream
stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

# Data Expansion (ADD EXACTLY CONTROL_NUM + 3)
custom_entries = []
for i in range(CONTROL_NUM + 3):
    campus = "TUP Manila" if i % 2 == 0 else "TUP Taguig"
    program = f"{STUDENT_MAJOR}_v{i}"
    custom_entries.append((campus, program))

stream.extend(custom_entries)

# Phase 1: Aggregation
campus_data = defaultdict(list)

for campus, program in stream:
    campus_data[campus].append(program)

# Phase 2: Frequency Analysis
manila_counter = Counter(campus_data["TUP Manila"])
top_program, top_count = manila_counter.most_common(1)[0]

# Store output 
ex1_output = {
    "CONTROL_NUM": CONTROL_NUM,
    "STUDENT_MAJOR": STUDENT_MAJOR,
    "Expanded Stream Length": len(stream),
    "TUP Manila Applications": len(campus_data["TUP Manila"]),
    "Top Program (Manila)": top_program,
    "Frequency": top_count
}
```

```python
# EXERCISE 2

# === REQUIRED VARIABLES ===
SEED_NUM = 1  # <-- CHANGE THIS
STUDENT_MAJOR = "BSME"  # MUST BE UPPERCASE
FAVORITE_ARTIST = "NICOLE"
LOCAL_HAZARD = "TYPHOON"

CONTROL_NUM = max(3, SEED_NUM)

# === EXERCISE 2 ===

# Festival lineup
festival = {
    "Ben&Ben", "SB19", "Bini", "Eraserheads",
    FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"
}

# Users
user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

# Data Expansion (ADD EXACTLY CONTROL_NUM artists)
for i in range(CONTROL_NUM):
    festival.add(f"Extra Artist {i}")

# Jaccard Function
def jaccard(a, b):
    intersection = a & b
    union = a | b
    return len(intersection), len(union), (len(intersection) / len(union)) * 100

# Compute
a_int, a_union, a_score = jaccard(user_a, festival)
b_int, b_union, b_score = jaccard(user_b, festival)

dealbreakers = user_a - festival

# Store output
ex2_output = {
    "Total Festival Artists": len(festival),
    "User A Intersection": a_int,
    "User A Union": a_union,
    "User A Similarity (%)": round(a_score, 2),
    "User B Similarity (%)": round(b_score, 2),
    "User A Dealbreakers": dealbreakers
}
```

```python
# EXERCISE 3

# === REQUIRED VARIABLES ===
SEED_NUM = 1  # <-- CHANGE THIS
STUDENT_MAJOR = "BSME"  # MUST BE UPPERCASE
FAVORITE_ARTIST = "NICOLE"
LOCAL_HAZARD = "TYPHOON"

CONTROL_NUM = max(3, SEED_NUM)

# === EXERCISE 3 ===
from collections import deque, Counter

dq = deque(maxlen=5)

# Base burst
burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

# Data Expansion (ADD EXACTLY CONTROL_NUM + 2 alerts)
start_time = CONTROL_NUM + 7
for i in range(CONTROL_NUM + 2):
    category = "WEATHER" if i % 2 == 0 else "TRAFFIC"
    burst.append((start_time + i, category))

# Tracking checks
weather_check = None
hazard_check = None

# Sliding Window Algorithm
for timestamp, category in burst:
    current_categories = [c for _, c in dq]
    freq = Counter(current_categories)

    if freq[category] >= 2:
        appended = False
    else:
        dq.append((timestamp, category))
        appended = True

    # Capture required checks
    if timestamp == CONTROL_NUM + 3:
        weather_check = appended

    if category == LOCAL_HAZARD:
        hazard_check = appended

# Final computation
timestamp_sum = sum(t for t, _ in dq)

# Store output
ex3_output = {
    "LOCAL_HAZARD": LOCAL_HAZARD,
    "Total Burst Size": len(burst),
    "WEATHER at CONTROL_NUM+3 appended": "Y" if weather_check else "N",
    "LOCAL_HAZARD appended": "Y" if hazard_check else "N",
    "Final Deque Length": len(dq),
    "Sum of Timestamps": timestamp_sum,
    "Final Deque": list(dq)
}
```
